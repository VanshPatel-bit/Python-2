### *WAP TO PRINT ELEMENTS IN LIST WITH FREQUENCY GREATER THAN A GIVEN VALUE K*

```python
l=[1,1,1,1,2,2,3,3,3,4,5,5,5,6,6]
k=int(input("ENTER THE VALUE OF K :"))
n=[]
for i in l:
    count=l.count(i)
    if count>k and i not in n:
        n.append(i)
print(n)
```

### *WAP TO REMOVE ALL ACCURANCE TO OF AN ELEMENT ENTERED BY USER FROM GIVEN LIST*

```python
l=[1,1,1,1,2,3,3,4,5,4,5,6]
k=int(input("ENTER THE ELEMENT TO REMOVE :"))
if k not in l:
    print("NUMBER NOT FOUND")
else:
    for i in l:
        count=l.count(k)
        if count>0:
            l.remove(k)
    print(l)
```

### *WAP that reads the word from the paragraph and stores them as keys in a dictonary and count the frequency of it as a*

```python
s="The that that that that boy used in that sentence was wrong."
words=s.split()
d={}
for word in words:
    if word in d:
        d[word]+=1
    else:
        d[word]=1
print(d)
```

### *WAP TO CREATE A DICTONARY WITH THE KEY AS THE FIRST CHARACTER OF THE WORD AND VALUE AS A LIST OF WORDS STARTING WITH THAT CHARACTER*

```python
s="The that that that that boy used in that sentence was wrong."
words=s.split()
p={}
for i in words:
    key=i[0]
    if key in p:
        p[key].append(i)
    else:
        p[key]=[i]
print(p)
```

### *WAP TO REPLACE THE WORDS OF A GIVEN STRING FROM THE DICTONARY*

```python
s="campusX best for ds students"
d={'best':'is the best channel','ds':'data science'}
l=[]
for i in s.split():
    if i in d:
        l.append(d[i])
    else:
        l.append(i)
x=" ".join(l)
print(x)
```

# *set*
1. MUTABLE
2. IF WE WANT TO REPRESENT A GROUP OF UNIQUE VALUE AS A SINGLE ENTITY THEN WE SHOULD GO FOR SET
3. DUPLICATES NOT ALLOWED
4. INSERTION ORDER NOT PRESERVED BUT WE CAN SORT THE ELEMENTS
5. INDEXING AND SLICING NOT ALLOWED FOR SET
6. HETROGENEOUS ELEMENTS ARE ALLOWED
7. WE CAN REPRESENT SET ELEMENTS WITHIN CURLY BRACKETS WITH COMMA SEPERATOR
8. WE CAN APPLY MATHEMATICAL OPERATIONS LIKE ~ UNION,INTERSECTION,DIFFERENCE ETC. ON SET

## MATHEMATICAL OPERATION OF SET

```python
# 1.UNION()
x={1,2,3,4,5,6,7}
y={5,6,7,4}
print(x.union(y))
print(x|y)
```

```python
# 2.INTERSECTION()
x={1,2,3,4,5,6,7}
y={5,6,7,4,9,10}
print(x.intersection(y))
print(x&x)
```

```python
# 3.DIFFERENCE()
x={1,2,3,4,5,6,7}
y={5,6,7,4,9,10}
print(x.difference(y))
print(x-y)
print(y-x)
```

```python
# 4.symmetric_difference()
x={1,2,3,4,5,6,7}
y={5,6,7,4,9,10}
print(x.symmetric_difference(y))
print(x^y)
print(y^x)
```

```python
# 5.issubset()
x={"a","b","c"}
y={"d","e","b","h","a","c"}
print(x.issubset(y))
print(y.issubset(x))
```

```python
# 6.issuperset()
x={"a","b","c"}
y={"d","e","b","h","a","c"}
print(x.issuperset(y))
print(y.issuperset(x))
```
