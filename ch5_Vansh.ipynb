###### Q. WAP that reads the word from the paragraph and stores them as keys in a dictionary and count the fequency of it as a...

```python
s="dog the quick brown fox jumps over the lazy dog"
words=s.split()
d={}
for word in words:
    if word in d:
        d[word]+=1
    else:
        d[word]=1
print(d)
```

##### Q. WAP to create a dictionary with the key as the first character of the word and value as a
#####    list of words starting with that character

```python
s="dog the quick that brown them fox jumps over the lazy dog"
words=s.split()
d={}
for word in words:
    first_char=word[0]
    if first_char in d:
        d[first_char].append(word)
    else:
        d[first_char]=[word]
print(d)
```

##### Q. WAP to replace the words of given string from the dictionary

```python
s="CampusX best for DS students"
d={'best':'is the best channel','DS':'Data Science'}
l=[]
for i in s.split():
    if i in d:
        l.append(d[i])
    else:
        l.append(i)
x=" ".join(l)
print(x)
```

# * Set
- mutable
- if we want to represent a group of unique value as a single entity then we should go for set
- duplicates not allowed
- insertion order not preserved but we can sort the elements
- indexing and slicing not allowed for set
- heterogeneous elements are allowed
- we can represent set elements within curly brackets with comma separator
- we can apply mathematical operation like Union,intersection,difference etc. on set

#### * Mathematical operation of set

```python
# 1. union()
x={10,20,30,40}
y={30,40,50,60}
print(x.union(y))
print(x|y)
```

```python
# 2. intersection()
x={10,20,30,40}
y={30,40,50,60}
print(x.intersection(y))
print(x&y)
```

```python
# 3. difference
x={10,20,30,40}
y={30,40,50,60}
print(x.difference(y))
print(x-y)
```

```python
# 4. symmetric_difference()
x={10,20,30,40}
y={30,40,50,60}
print(x.symmetric_difference(y))
print(x^y)
```

```python
# 5. issubset()
x={"a","b","c"}
y={"d","e","a","h","b","c"}
print(x.issubset(y))
print(y.issubset(x))
```

```python
# 6. issuperset()
x={"a","b","c"}
y={"d","e","a","h","b","c"}
print(x.issuperset(y))
print(y.issuperset(x))
```

### Lambda Function
- we can declare a function without any name,such type of nameless function is called anonymous function or lambda function
- #### Syntax
- Lambda argument_list : expression 

```python
def square(n):
    return n*n
print(square(5))
```

```python
s=lambda x:x*x
print(s(5))
```

```python
s=lambda x,y:x+y
print(s(10,20))
```

### map() function
- For every element present in the given sequence,apply some functionality and generate new element with required modification
- #### Syntax
- **map(function,sequence)**
- The function can be applied on each element of sequence and generate new sequence

```python
l=[1,2,3,4,5]
def double(x):
    return x*2
l1=list(map(double,l))
print(l1)
```

```python
l=[1,2,3,4,5]
l1=list(map(lambda x:x*2,l))
print(l1)
```

```python
l1=[1,2,3,4,5]
l2=[4,5,6,7,8]
l3=[]
for i in range(len(l1)):
    l3.append(l1[i]+l2[i])
print(l3)
```

```python
l1=[1,2,3,4,5]
l2=[4,5,6,7,8]
l3=list(map(lambda x,y:x+y,l1,l2))
print(l3)
```

### reduce() Function
- reduces sequence of elements into a single element by appling the specified function.
- #### Syntax
- **reduce(function,sequence)**
- reduce() function present in functools module and hence we need to import the module first to use the reduce() function.

```python
import math
print(math.sqrt(9))
```

```python
from math import *
print(sqrt(9))
```

```python
from functools import *
l=[1,2,3,4,5]
sum=reduce(lambda x,y:x*y,l)
print(sum)
```

### filter() function
- used to filter values from the given sequence based on some condition
- ### Syntax
- **filter(function,sequence)**
- where function argument responsible to perform conditional check and sequence can be list or tuple or string.

```python
l=[10,25,20,22,40,35,45,50]
def iseven(x):
    if x%2==0:
        return True
    else:
        return False
l1=list(filter(iseven,l))
print(l1)
```

```python
l=[10,25,20,22,40,35,45,50]
l1=list(filter(lambda x:x%2==1,l))
print(l1)
```
