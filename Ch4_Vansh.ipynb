```python
s="Py$2 37@th42 3on"
alphabets=0
upper=0
lower=0
digit=0
special=0
space=0

for x in s:
    if x.isalpha():
        alphabets+=1
        if x.isupper():
            upper+=1
        elif x.islower():
            lower+=1
    elif x.isdigit():
        digit+=1
    elif x.isspace():
        space+=1
    else:
        special+=1

print("Total Alphabets : ",alphabets)
print("Upper case : ",upper)
print("Lower case : ",lower)
print("Total digit : ",digit)
print("Total space : ",space)
print("Total special : ",special)
```

### WAP to check the password entered by user is valid or not base on following condition
##### 1. Minimum 8 character
##### 2. Alphabets must be between(a-z)
##### 3. Atleast one uppercase
##### 4. Atleast one digit
##### 5. Atleast one special character from(_,$,@)

```python
import string 
pw=input("Enter Password : ")
l,u,sp,d=0,0,0,0
if len(pw)>=8:
    for i in pw:
        if i.isupper():
            u+=1
        if i.islower():
            l+=1
        if i.isdigit():
            d+=1
        if i=="_" or i=="$" or i=="@":
            sp+=1
    if l>0 and u>0 and d>0 and sp>0 and (u+l+sp+d)==len(pw):
        print("Valid Password")
    else:
        print("Unvalid Password")
else:
    print("Invalid")
```

```python
email=input("Enter a Email : ")
if '@' not in email:
    print("Invalid email")
else:
    s=email.split('@')
    username=s[0]
    domain=s[1]
    if len(username)>1:
        masked_username=username[0]+'*'*(len(username)-1)
    else:
        masked_username=username
    x=masked_username+'@'+domain
    print(email)
    print(x)
```
